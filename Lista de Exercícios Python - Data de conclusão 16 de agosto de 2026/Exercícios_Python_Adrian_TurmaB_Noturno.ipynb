21. Busca Binária, em lista ordenada, comparo com o elemento central, se for igual achei, se menor procuro na metade esquerda, se maior na direita e repito até achar ou acabar o intervalo é muito mais rápido que percorrer tudo.

```python
def busca(lista, valor):
    inicio = 0
    fim = len(lista) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2
        if lista[meio] == valor:
            return meio
        elif lista[meio] < valor:
            inicio = meio + 1
        else:
            fim = meio - 1
    return -1


dados = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
alvo = int(input("Digite o número que quer buscar: "))

posicao = busca(dados, alvo)
if posicao != -1:
    print("Encontrado na posição", posicao)
else:
    print("Não encontrado")
```

22. Criptografia de Cifra de César, converti letra para número, desloquei, voltei para letra, eu usei ord() e chr() para converter entre letra e código numérico o % 26 faz voltar ao início do alfabeto quando passa da letra Z e mantive maiúsculas e minúsculas e não alterei símbolos.

```python
def criptografar(texto, deslocamento):
    resultado = ""
    for letra in texto:
        if letra.isalpha():
            if letra.isupper():
                base = ord('A')
            else:
                base = ord('a')
            nova = chr((ord(letra) - base + deslocamento) % 26 + base)
            resultado = resultado + nova
        else:
            resultado = resultado + letra
    return resultado

mensagem = input("Digite a mensagem: ")
chave = int(input("Deslocamento: "))

print("Cifrado:", criptografar(mensagem, chave))
```

23.  Descriptografar Cifra de César, descriptografar é deslocar para trás, aproveitei a mesma função e passei deslocamento negativo e assim não repito código.

```python
def descriptografar(texto, deslocamento):
    return criptografar(texto, -deslocamento)

cifrado = input("Digite o texto cifrado: ")
chave = int(input("Deslocamento usado: "))

print("Original:", descriptografar(cifrado, chave))
```

24. Validação de CPF, eu removi caracteres não numéricos e verifiquei tamanho e se todos os dígitos são iguais caso inválido também calculei o primeiro e o segundo dígito verificador conforme regra oficial e comparei com o informado.

```python
def validar_cpf(cpf):
    cpf = ''.join([c for c in cpf if c.isdigit()])

    if len(cpf) != 11:
        return False

    if cpf == cpf[0] * 11:
        return False

    soma = 0
    for i in range(9):
        soma = soma + int(cpf[i]) * (10 - i)
    d1 = 0 if soma % 11 < 2 else 11 - (soma % 11)

    soma = 0
    for i in range(10):
        soma = soma + int(cpf[i]) * (11 - i)
    d2 = 0 if soma % 11 < 2 else 11 - (soma % 11)

    return cpf[-2:] == str(d1) + str(d2)


cpf_informado = input("Digite o CPF: ")

if validar_cpf(cpf_informado):
    print("CPF válido")
else:
    print("CPF inválido")
```

31. Ordenação por Bolha, eu percorro a lista comparando pares vizinhos se estiverem fora de ordem troco de posição, repito até uma passagem completa sem precisar trocar nada, sinal de que está ordenada e a cada volta o maior elemento borbulha para o fim.

```python
def ordenacao_bolha(lista):
    n = len(lista)
    for i in range(n):
        for j in range(n - i - 1):
            if lista[j] > lista[j + 1]:
                # Troca de posição
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
    return lista

dados = [6, 3, 9, 1, 4, 2]
print("Lista ordenada:", ordenacao_bolha(dados))
```

```python
tabuleiro = [" "] * 9
jogador_atual = "X"

def exibir_tabuleiro():
    print(tabuleiro[0], "|", tabuleiro[1], "|", tabuleiro[2])
    print("---------")
    print(tabuleiro[3], "|", tabuleiro[4], "|", tabuleiro[5])
    print("---------")
    print(tabuleiro[6], "|", tabuleiro[7], "|", tabuleiro[8])

def verificar_vitoria():
    linhas = [(0,1,2),(3,4,5),(6,7,8),
              (0,3,6),(1,4,7),(2,5,8),
              (0,4,8),(2,4,6)]
    for a,b,c in linhas:
        if tabuleiro[a] == tabuleiro[b] == tabuleiro[c] != " ":
            return True
    return False

for _ in range(9):
    exibir_tabuleiro()
    pos = int(input(f"Jogador {jogador_atual}, escolha posição (1-9): ")) - 1

    if tabuleiro[pos] == " ":
        tabuleiro[pos] = jogador_atual

        if verificar_vitoria():
            exibir_tabuleiro()
            print(f"Jogador {jogador_atual} venceu!")
            break

        jogador_atual = "O" if jogador_atual == "X" else "X"
    else:
        print("Posição já ocupada!")
else:
    exibir_tabuleiro()
    print("Deu velha!")
```

46. Calcular Média e Desvio Padrão, eu calculei a média, depois calculei a diferença de cada valor para a média, elevei ao quadrado, somei tudo, dividi pela quantidade e tirei raiz quadrada, eu segui a fórmula passo a passo.

```python
from math import sqrt

dados = [10, 12, 15, 18, 20]

media = sum(dados) / len(dados)

soma_dos_quadrados = 0
for valor in dados:
    soma_dos_quadrados = soma_dos_quadrados + (valor - media) ** 2

desvio_padrao = sqrt(soma_dos_quadrados / len(dados))

print("Média dos valores:", media)
print("Desvio padrão:", desvio_padrao)
```

48. Ordenação por Inserção, peguei cada elemento e o insiri na posição correta dentro da porção já ordenada a esquerda e vou empurrando os elementos maiores uma posição a frente até achar o lugar certo.

```python
def ordenacao_insercao(lista):
    for i in range(1, len(lista)):
        chave = lista[i]
        j = i - 1

        while j >= 0 and lista[j] > chave:
            lista[j + 1] = lista[j]
            j = j - 1

        lista[j + 1] = chave
    return lista

dados = [8, 2, 5, 1, 9, 3]
print("Lista ordenada:", ordenacao_insercao(dados))
```

49. Soma dos Dígitos, eu li o número como texto para percorrer cada caractere, ai eu converti cada dígito de volta para inteiro e fui somando, desse jeito não precisei usar divisão.

```python
numero = input("Digite um número inteiro: ")

soma = 0
for digito in numero:
    soma = soma + int(digito)

print("Soma dos dígitos:", soma)
```

50. Cálculo de Frequência de Letras, eu filtrei apenas letras com .isalpha(), e converti tudo para minúsculo para não diferenciar maiúsculas de minúsculas na contagem.

```python
texto = input("Digite um texto: ").lower()

frequencia = {}
for letra in texto:
    if letra.isalpha():
        if letra in frequencia:
            frequencia[letra] = frequencia[letra] + 1
        else:
            frequencia[letra] = 1

print("Frequência de cada letra:")
for l, qtd in frequencia.items():
    print(l, "→", qtd)
```
